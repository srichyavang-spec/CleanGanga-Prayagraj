For BOD we have one clear threshold:
BOD ≤ 3 mg/L       → within criterion
BOD > 3 mg/L       → exceeds criterion

For Fecal Coliform:
FC ≤ 500            → desirable
500 < FC ≤ 2500     → above desirable, but within maximum permissible
FC > 2500           → exceeds maximum permissible

# CPCB Class B bathing-water criteria
BOD_LIMIT = 3                  # mg/L
FC_DESIRABLE = 500             # MPN/100 mL
FC_MAX = 2500                  # MPN/100 mL

# Check whether each observation exceeds the BOD criterion.
prayagraj["BOD_exceeds"] = (prayagraj["Biochemical Oxygen Demand (mg/L)"] >BOD_LIMIT)

# Check whether Fecal Coliform exceeds the desirable level.
prayagraj["FC_above_desirable"] = ( prayagraj["Fecal Coliform (MPN/100ml)"] > FC_DESIRABLE)

# Check whether Fecal Coliform exceeds the maximum permissible level.
prayagraj["FC_exceeds_max"] = (prayagraj["Fecal Coliform (MPN/100ml)"] > FC_MAX)

# Display the result so we can inspect individual observations.
display(prayagraj[["Station","Biochemical Oxygen Demand (mg/L)","Fecal Coliform (MPN/100ml)","BOD_exceeds","FC_above_desirable","FC_exceeds_max"]])
